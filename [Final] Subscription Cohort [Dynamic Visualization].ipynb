- Import required libraries

```python
# Importing the required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

- Import dataset

```python
df = pd.read_csv('customer_behavior_history.csv', parse_dates=['signup_date', 'event_timestamp'])
df = df.drop(columns=['key','device_id', 'device_category', 'signup_date', 'watch_time', 'amount'])
df.tail()
```

- Keep only retention event

```python
df = df[(df['event_type'] == 'billing') & (df['event_status']== 'Success')].copy()
df.tail()
```

- Create a cohort group for each customer based on their first purchase date
- Transform the event_timestamp to extract the month period

```python
df['first_event'] = (
    df.groupby('member_id')['event_timestamp']
    .transform('min')
)

df['cohort_first_month'] = (
    df['first_event']
    .dt.to_period('M')
)

df.tail()
```

- Transform the event_timestamp to extract the month period

```python
df['first_event'] = (
    df.groupby('member_id')['event_timestamp']
    .transform('min')
)

df['cohort_first_month'] = (
    df['first_event']
    .dt.to_period('M')
)
df['event_month'] = df['event_timestamp'].dt.to_period('M')
df = df[(df['event_month'] <= pd.Period('2025-12', freq='M')) 
            & (df['event_month'] >= pd.Period('2025-01', freq='M'))
            & (df['cohort_first_month'] <= pd.Period('2025-12', freq='M')) 
            & (df['cohort_first_month'] >= pd.Period('2025-01', freq='M'))]
df.tail()
```

```python
df.dtypes
```

- Create the number of retention months

```python

df['month_diff'] = (
    df['event_month'] - df['cohort_first_month']
).apply(lambda x: x.n)


df.tail()
```

```python
cohort_size = df[df['month_diff'] == 0].groupby('cohort_first_month')['member_id'].nunique().rename('cohort_size')
```

- Similar pratice in Looker Studio (Data Studio)

```python
cohort_data = (
    df.groupby(['cohort_first_month', 'month_diff'])['member_id']
    .nunique()
    .reset_index()
    .rename(columns={'member_id': 'retained_users'})
)

cohort_data = cohort_data.join(cohort_size, on='cohort_first_month')
cohort_data['retention_rate'] = (
    cohort_data['retained_users'] / cohort_data['cohort_size'] * 100
).round(1)
```

```python
cohort_data
```

```python
cohort_pivot_percent = cohort_data.pivot_table(
    index='cohort_first_month',
    columns='month_diff',
    values='retention_rate'
)
cohort_pivot_percent.index = cohort_pivot_percent.index.astype(str)
```

```python
cohort_pivot_absolute = cohort_data.pivot_table(
    index='cohort_first_month',
    columns='month_diff',
    values='retained_users'
    )
cohort_pivot_absolute.index = cohort_pivot_absolute.index.astype(str)
```

- Visualization

```python
def plot_dynamic_heatmap(data, is_percent=False, title=None):
    """
    Create an interactive Plotly heatmap for cohort analysis
    """
    # Convert data to suitable format for Plotly
    z_values = data.values
    x_values = data.columns.astype(str)
    y_values = data.index.astype(str)
    
    # Choose colorscale based on type
    colorscale = 'Blues' if is_percent else 'Blues'
    
    # Create hover text
    hover_text = []
    for i, row in enumerate(z_values):
        hover_row = []
        for j, val in enumerate(row):
            if is_percent:
                hover_row.append(f"Cohort: {y_values[i]}<br>Month: {x_values[j]}<br>Retention: {val:.1f}%")
            else:
                hover_row.append(f"Cohort: {y_values[i]}<br>Month: {x_values[j]}<br>Users: {int(val)}")
        hover_text.append(hover_row)
    
    fig = go.Figure(data=go.Heatmap(
        z=z_values,
        x=x_values,
        y=y_values,
        colorscale=colorscale,
        hovertext=hover_text,
        hoverinfo='text',
        text=z_values,
        texttemplate='%{text:.1f}' if is_percent else '%{text:.0f}',
        textfont={"size": 11, "color": "black"},
        showscale=True
    ))
    
    # Update layout for dark theme
    fig.update_layout(
        title=title if title else ('Retention Rate Heatmap (%)' if is_percent else 'User Count Heatmap'),
        title_font=dict(size=16, color='white'),
        xaxis_title='Period (Months Since First Purchase)',
        yaxis_title='Cohort Month',
        plot_bgcolor='#1e1e1e',
        paper_bgcolor='#1e1e1e',
        font=dict(color='white', size=12),
        xaxis=dict(
            side='top',
            tickfont=dict(color='white', size=10),
            title_font=dict(color='white', size=12)
        ),
        yaxis=dict(
            tickfont=dict(color='white', size=10),
            title_font=dict(color='white', size=12)
        ),
        height=500,
        width=1000,
        margin=dict(l=100, r=50, t=80, b=50)
    )
    
    # Update colorbar
    fig.update_traces(
        colorbar=dict(
            title='Retention Rate (%)' if is_percent else 'Users',
            title_font=dict(color='white', size=12),
            tickfont=dict(color='white', size=10)
        )
    )
    
    return fig

def plot_dynamic_cohort_curve(data):
    """
    Create an interactive Plotly line chart for cohort retention curves
    """
    fig = go.Figure()
    
    # Generate blue color palette
    colors = px.colors.sequential.Blues
    
    for i, cohort in enumerate(data.index):
        fig.add_trace(go.Scatter(
            x=data.columns.astype(str),
            y=data.loc[cohort].values,
            mode='lines+markers',
            name=str(cohort),
            line=dict(
                color=colors[i % len(colors)],
                width=2
            ),
            marker=dict(
                size=6,
                color=colors[i % len(colors)]
            ),
            hovertemplate='<b>%{fullData.name}</b><br>' +
                         'Month: %{x}<br>' +
                         'Retention: %{y:.1f}%<br>' +
                         '<extra></extra>'
        ))
    
    # Update layout for dark theme
    fig.update_layout(
        title='Retention Curves by Cohort',
        title_font=dict(size=16, color='white'),
        xaxis_title='Months Since First Purchase',
        yaxis_title='Retention Rate',
        plot_bgcolor='#1e1e1e',
        paper_bgcolor='#1e1e1e',
        font=dict(color='white', size=12),
        xaxis=dict(
            tickfont=dict(color='white', size=10),
            title_font=dict(color='white', size=12),
            gridcolor='#2a2a2a',
            showgrid=True
        ),
        yaxis=dict(
            tickfont=dict(color='white', size=10),
            title_font=dict(color='white', size=12),
            gridcolor='#2a2a2a',
            showgrid=True,
            tickformat='.0%'
        ),
        height=500,
        width=1000,
        margin=dict(l=100, r=150, t=80, b=50),
        legend=dict(
            title=dict(text='Cohort', font=dict(color='white', size=12)),
            font=dict(color='white', size=10),
            bgcolor='rgba(30, 30, 30, 0.8)',
            bordercolor='#2a2a2a',
            borderwidth=1,
            x=1.02,
            y=1,
            xanchor='left',
            yanchor='top'
        ),
        hovermode='x unified'
    )
    
    return fig
```

```python
# Create dynamic interactive visualizations using Plotly

# Dynamic heatmap for absolute user counts
fig_absolute = plot_dynamic_heatmap(cohort_pivot_absolute, is_percent=False, title='User Count Heatmap')
fig_absolute.show()

# Dynamic heatmap for retention rates
fig_percent = plot_dynamic_heatmap(cohort_pivot_percent, is_percent=True, title='Retention Rate Heatmap (%)')
fig_percent.show()

# Dynamic cohort retention curves
fig_curves = plot_dynamic_cohort_curve(cohort_pivot_percent)
fig_curves.show()
```
